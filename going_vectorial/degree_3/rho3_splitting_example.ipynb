```python
from sympy import *

v1, w1 = Matrix([-sqrt(2)+1, 0, sqrt(2)+1, 0]), Matrix([0, sqrt(3)+sqrt(2), 0, -1])
v2, w2 = Matrix([+sqrt(2)+1, 0, sqrt(2)-1, 0]), Matrix([0, sqrt(3)+sqrt(2), 0, +1])
Matrix([v1.T, w1.T])
```

```python
Matrix([v2.T, w2.T])
```

```python
v1 = (v1 / sqrt(v1.dot(v1))).expand()
w1 = (w1 / sqrt(w1.dot(w1))).expand()
v2 = (v2 / sqrt(v2.dot(v2))).expand()
w2 = (w2 / sqrt(w2.dot(w2))).expand()

Matrix([v1.T, w1.T])
```

```python
Matrix([v2.T, w2.T])
```

```python
assert v1.dot(v1).equals(1) and w1.dot(w1).equals(1) and v1.dot(w1).equals(0)
assert v2.dot(v2).equals(1) and w2.dot(w2).equals(1) and v2.dot(w2).equals(0)
```

```python
from cq.symbolic import states_to_density, rho_to_g

rho1 = states_to_density(v1, w1).expand()
rho2 = states_to_density(v2, w2).expand()
g1 = rho_to_g(rho1)
g2 = rho_to_g(rho2)
g1.simplify()
g2.simplify()

g1
```

```python
g2
```

```python
assert g1.equals(g2)
```

```python
from cq.symbolic import rho_to_T

T1 = rho_to_T(rho1).expand()
T2 = rho_to_T(rho2).expand()
T1
```
